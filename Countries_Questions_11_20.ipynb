### **First Import The Libraries**

```python
import numpy as np
import pandas as pd
```

### **Read CSV**

```python
data = pd.read_csv(r'Countries.csv')
```

### **Data Overview (info & describe)**

```python
# info() shows column names, data types, and non-null counts — useful to spot missing data and dtype issues
data.info()
```

```python
# describe() gives summary statistics (mean, min, max, std, etc.) for all numeric columns
data.describe()
```

**Question 11 : Which continent has the highest number of countries?**

```python
# Group rows by continent, count countries in each group, then nlargest(1) picks the top group

data.groupby('continent')['country'].count().nlargest(1)
```

**Question 12 : Which country has the highest GDP**

```python
# idxmax() finds the row index with the highest gdp, then .loc picks the country at that index

data.loc[data['gdp'].idxmax() , 'country']
```

**Question 13 : Which country has the lowest GDP**

```python
# Filter rows where gdp equals the minimum gdp value, then pick the country column

data[data['gdp'] == data['gdp'].min()]['country']
```

**Question 14 : What is the average population of all countries**

```python
# mean() calculates the average population, round(2) keeps it to 2 decimal places

print(data['population'].mean().round(2))
```

**Question 15 : Which country has the highest life expectancy**

```python
# idxmax() finds the row index with the highest life_expectancy, then .loc picks that country

data.loc[data['life_expectancy'].idxmax(), 'country']
```

**Question 16 : Which country has the lowest life expectancy**

```python
# idxmin() finds the row index with the lowest life_expectancy, then .loc picks that country

data.loc[data['life_expectancy'].idxmin() , 'country']
```

**Question 17 : How many countries have 100% electricity access**

```python
# Check where electricity_access_pct equals 100, sum() counts how many rows are True

print((data['electricity_access_pct'] == 100).sum())
```

**Question 18 : Which country has the highest internet penetration percentage**

```python
# Filter rows where internet_pct equals its max value, then pick the country column

data[data['internet_pct'] == data['internet_pct'].max()]['country']
```

**Question 19 : Which continent has the highest average life expectancy**

```python
# Group rows by continent, take the mean life_expectancy per group, nlargest(1) picks the top one

data.groupby('continent')['life_expectancy'].mean().nlargest(1)
```

**Question 20 : How many countries have a population greater than 100 million**

```python
# Check where population is greater than 100 million, sum() counts how many rows are True

print((data['population'] > 100000000).sum())
```
