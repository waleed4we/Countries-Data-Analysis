### **First Import The Libraries**

```python
import numpy as np
import pandas as pd
```

### **Read CSV**

```python
data = pd.read_csv(r'Countries.csv')
```

### **Data Overview (info & describe)**

```python
# info() shows column names, data types, and non-null counts — useful to spot missing data and dtype issues
data.info()
```

```python
# describe() gives summary statistics (mean, min, max, std, etc.) for all numeric columns
data.describe()
```

**Question 1 : What Is The Capital Of The Country With Highest Population**

```python
# Find the row where population is at its max, then pick the capital_city column

data[data['population'] == data['population'].max()]['capital_city']
```

**Question 2 : Which Country Has Least Population**

```python
# Find the row where population is at its min , then pick the country_long column for country name

data[data['population'] == data['population'].min()]['country_long']
```

**Question 3 : What Is Capital Of Least Populated Country**

```python
# Same least-population row, but this time extract the capital_city column

data[data['population'] == data['population'].min()]['capital_city']
```

**Question 4 : Top 5 Countries With Highest Democratic Score**

```python
# nlargest() sorts and picks top 5 rows by democracy_score, then we grab the country names

print(data.nlargest(5 , 'democracy_score')['country'])
```

**Question 5 : How Many Total Regions Are There**

```python
# value_counts() groups rows by region, count() then tells us how many unique regions exist

print(data['region'].value_counts().count())   
```

**Question 6 : How Many Countries Lie In Eastern Europe Region**

```python
# For Names
# Filter rows where region equals 'Eastern Europe' and list their country names

data[data['region'] == "Eastern Europe"][ 'country']
```

```python
# For Only Count

print(data[data['region'] == 'Eastern Europe']['country'].count())
```

**Question 7 : Who Is The Political Leader Of Second Highest Populated Country**

```python
# nlargest(2) gets the top 2 populations, .iloc[1] picks the 2nd highest value,
# then we filter the DataFrame to find that country's political_leader

data[data['population'] == data['population'].nlargest(2).iloc[1]]['political_leader']
```

**Question 8 :  How Many Countries Are There Whoes Political Leaders Are Unknown**

```python
# isna() flags rows where political_leader is missing (NaN), count() totals them up

print(data[data['political_leader'].isna()]['country'].count())
```

**Question 9 : How Many Countries Have Republic In Their Full Name**

```python
# Define a counter and a function that checks if 'republic' appears in the country's full name
# (case-insensitive), incrementing the global count each time it's found

count = 0
def republic_count(full_name):
    global count
    if pd.notna(full_name) and 'republic' in full_name.lower():
        count += 1
    return count
```

```python
# apply() runs the function on every row of country_long, then we print the final count

data['country_long'].apply(republic_count)
count
```

**Question 10 : Which Country In African Region Has Highest Population**

```python
# Filter the DataFrame to only African countries , then find the one with max population

african_df = data[data['continent'] == 'Africa']
print(african_df[african_df['population'] == african_df['population'].max()]['country'])
```
